1. Forward Propagation:

*   Hidden Layer: $Z_h = x \cdot W_h + \theta_h = [-0.7, 0.1]$, resulting in $A_h = [0.0, 0.1]$ after applying ReLU.
*   Output Layer: $Z_o = A_h \cdot W_o + \theta_o = 0.08$, resulting in $\hat{y} = 0.08$ after applying ReLU.
* Loss: $L = \frac{1}{2} (\hat{y} - y)^2 = 0.4232$.

2. Backward Propagation:

* Calculates the error derivative at the output: $d_o = (\hat{y} - y) \cdot f'(Z_o) = -0.92 \cdot 1 = -0.92$.
* Propagates error back to hidden layer: $d_h = (d_o \cdot W_o) \circ f'(Z_h)$.* Updates weights via gradient descent: $W \leftarrow W - \text{lr} \times \nabla W$.

```python
import numpy as np

# 1. Inputs and Parameters
x = np.array([1, 0, 1])
y = np.array([1])
lr = 0.001

W_h = np.array([
    [0.2, -0.3],
    [0.4,  0.1],
    [-0.5, 0.2]
])
W_o = np.array([-0.3, -0.2])
theta_h = np.array([-0.4, 0.2])
theta_o = 0.1

# 2. Activation Functions
def relu(z):
    return np.maximum(0, z)

def relu_derivative(z):
    return (z > 0).astype(float)

# 3. Forward Pass
Z_h = np.dot(x, W_h) + theta_h
A_h = relu(Z_h)

Z_o = np.dot(A_h, W_o) + theta_o
y_hat = relu(Z_o)

loss = 0.5 * (y_hat - y)**2

# 4. Backward Pass (Backpropagation)
dLoss_dy_hat = y_hat - y
d_o = dLoss_dy_hat * relu_derivative(Z_o)

dW_o = A_h * d_o
dtheta_o = d_o

d_h = (d_o * W_o) * relu_derivative(Z_h)

dW_h = np.outer(x, d_h)
dtheta_h = d_h

# 5. Weight & Bias Updates
W_o_updated = W_o - lr * dW_o
theta_o_updated = theta_o - lr * dtheta_o

W_h_updated = W_h - lr * dW_h
theta_h_updated = theta_h - lr * dtheta_h

# 6. Print Results (.item() cleanly prevents DeprecationWarnings)
print("--- Forward Pass Results ---")
print(f"Predicted Output (y_hat) : {y_hat.item():.4f}")
print(f"Loss                     : {loss.item():.4f}")

print("\n--- Backward Pass Gradients ---")
print(f"Output Weight Gradients (dW_o)  :\n{dW_o}")
print(f"Hidden Weight Gradients (dW_h)  :\n{dW_h}")

print("\n--- Updated Parameters ---")
print(f"Updated Output Weights (W_o)    :\n{W_o_updated}")
print(f"Updated Hidden Weights (W_h)    :\n{W_h_updated}")
```
