## Laboratory Task 4: PyTorch Linear Regression

**Instruction:** Train a linear regression model in PyTorch using a regression dataset. Use the following parameters:

* **Criterion:** MSE Loss
* **Fully Connected Layers:** x 2
* **Batch Size:** 8
* **Optimizer:** SGD
* **Epochs:** 1000

```python
import torch
import torch.nn as nn
import torch.optim as optim
from torch.utils.data import TensorDataset, DataLoader

# 1. Synthetic Regression Dataset Generation
torch.manual_seed(42)
X = torch.randn(100, 4)  # 100 samples, 4 features
true_weights = torch.tensor([[2.0], [-1.5], [3.0], [0.5]])
true_bias = 1.2
y = torch.matmul(X, true_weights) + true_bias + torch.randn(100, 1) * 0.1

# Prepare DataLoader (Batch Size: 8)
dataset = TensorDataset(X, y)
dataloader = DataLoader(dataset, batch_size=8, shuffle=True)

# 2. Model Architecture (Fully Connected Layers x 2)
class TwoLayerRegressionModel(nn.Module):
    def __init__(self, input_dim):
        super(TwoLayerRegressionModel, self).__init__()
        self.fc1 = nn.Linear(input_dim, 8)  # Hidden Layer
        self.relu = nn.ReLU()
        self.fc2 = nn.Linear(8, 1)          # Output Layer

    def forward(self, x):
        out = self.fc1(x)
        out = self.relu(out)
        out = self.fc2(out)
        return out

model = TwoLayerRegressionModel(input_dim=4)

# 3. Criterion & Optimizer Setup
criterion = nn.MSELoss()                           # Criterion: MSE Loss
optimizer = optim.SGD(model.parameters(), lr=0.01) # Optimizer: SGD

# 4. Training Loop (Epoch: 1000)
epochs = 1000
print("--- Starting Training Loop ---")

for epoch in range(1, epochs + 1):
    running_loss = 0.0
    for batch_X, batch_y in dataloader:
        # Zero gradients
        optimizer.zero_grad()

        # Forward pass
        outputs = model(batch_X)
        loss = criterion(outputs, batch_y)

        # Backward pass & update
        loss.backward()
        optimizer.step()

        running_loss += loss.item() * batch_X.size(0)

    epoch_loss = running_loss / len(dataset)

    # Print loss every 100 epochs
    if epoch % 100 == 0 or epoch == 1:
        print(f"Epoch [{epoch}/{epochs}] - MSE Loss: {epoch_loss:.6f}")

print("\n--- Training Complete ---")
```
