# PyTorch CNN Architecture Implementation

This document provides the PyTorch implementation and detailed dimension calculations for the CNN architecture.

## 1. Spatial Dimension Calculations

Below are the step-by-step tensor shape transformations through the network layers:

- **Input**: $(1, 28, 28)$ (Channels: 1, Height: 28, Width: 28)
- **Conv1 + ReLU**:
  - Parameters: `kernel=3`, `stride=1`, `padding=1`
  - Output shape: `(32, 28, 28)`
- **MaxPool1**:
  - Parameters: `kernel=2`, `stride=2`, `padding=1`
  - Calculation: $\lfloor \frac{28 - 2 + 2(1)}{2} \rfloor + 1 = 15$
  - Output shape: `(32, 15, 15)`
- **Conv2 + ReLU**:
  - Parameters: `kernel=3`, `stride=1`, `padding=1`
  - Output shape: `(64, 15, 15)`
- **Conv3 + ReLU**:
  - Parameters: `kernel=3`, `stride=1`, `padding=1`
  - Output shape: `(128, 15, 15)`
- **Conv4 + ReLU**:
  - Parameters: `kernel=3`, `stride=1`, `padding=1`
  - Output shape: `(256, 15, 15)`
- **MaxPool2**:
  - Parameters: `kernel=2`, `stride=2`, `padding=0`
  - Calculation: $\lfloor \frac{15 - 2}{2} \rfloor + 1 = 7$
  - Output shape: `(256, 7, 7)`
- **Flatten**:
  - Shape: $(256 \times 7 \times 7) = 12,544$ features
- **Fully Connected Layers (FCN)**:
  - **FCN1**: `in_features=12544`, `out_features=1000`
  - **FCN2**: `in_features=1000`, `out_features=500`
  - **FCN3**: `in_features=500`, `out_features=num_classes` (e.g., 10)

## 2. PyTorch Code Implementation

```python
import torch
import torch.nn as nn
import torch.nn.functional as F

class CNNArchitecture(nn.Module):
    def __init__(self, num_classes=10):
        super(CNNArchitecture, self).__init__()
        
        # Conv1 Block
        self.conv1 = nn.Conv2d(in_channels=1, out_channels=32, kernel_size=3, stride=1, padding=1)
        self.pool1 = nn.MaxPool2d(kernel_size=2, stride=2, padding=1)
        
        # Conv2, Conv3, Conv4 Block
        self.conv2 = nn.Conv2d(in_channels=32, out_channels=64, kernel_size=3, stride=1, padding=1)
        self.conv3 = nn.Conv2d(in_channels=64, out_channels=128, kernel_size=3, stride=1, padding=1)
        self.conv4 = nn.Conv2d(in_channels=128, out_channels=256, kernel_size=3, stride=1, padding=1)
        self.pool2 = nn.MaxPool2d(kernel_size=2, stride=2, padding=0)
        
        # Dropout
        self.dropout = nn.Dropout(p=0.2)
        
        # Fully Connected Layers
        # Flattened dimension = 256 * 7 * 7 = 12544
        self.fcn1 = nn.Linear(in_features=12544, out_features=1000)
        self.fcn2 = nn.Linear(in_features=1000, out_features=500)
        self.fcn3 = nn.Linear(in_features=500, out_features=num_classes)

    def forward(self, x):
        # Layer 1
        x = F.relu(self.conv1(x))
        x = self.pool1(x)
        
        # Layers 2 - 4
        x = F.relu(self.conv2(x))
        x = F.relu(self.conv3(x))
        x = F.relu(self.conv4(x))
        x = self.pool2(x)
        
        # Dropout & Flatten
        x = self.dropout(x)
        x = torch.flatten(x, start_dim=1)
        
        # Fully Connected Layers
        x = F.relu(self.fcn1(x))
        x = F.relu(self.fcn2(x))
        x = F.softmax(self.fcn3(x), dim=1)
        
        return x
```

````python
import torch
import torch.nn as nn
import torch.nn.functional as F

class CNNArchitecture(nn.Module):
    def __init__(self, num_classes=10):
        super(CNNArchitecture, self).__init__()

        # Conv1 Block
        self.conv1 = nn.Conv2d(in_channels=1, out_channels=32, kernel_size=3, stride=1, padding=1)
        self.pool1 = nn.MaxPool2d(kernel_size=2, stride=2, padding=1)

        # Conv2, Conv3, Conv4 Block
        self.conv2 = nn.Conv2d(in_channels=32, out_channels=64, kernel_size=3, stride=1, padding=1)
        self.conv3 = nn.Conv2d(in_channels=64, out_channels=128, kernel_size=3, stride=1, padding=1)
        self.conv4 = nn.Conv2d(in_channels=128, out_channels=256, kernel_size=3, stride=1, padding=1)
        self.pool2 = nn.MaxPool2d(kernel_size=2, stride=2, padding=0)

        # Dropout
        self.dropout = nn.Dropout(p=0.2)

        # Fully Connected Layers
        # Flattened dimension = 256 * 7 * 7 = 12544
        self.fcn1 = nn.Linear(in_features=12544, out_features=1000)
        self.fcn2 = nn.Linear(in_features=1000, out_features=500)
        self.fcn3 = nn.Linear(in_features=500, out_features=num_classes)

    def forward(self, x):
        # Layer 1
        x = F.relu(self.conv1(x))
        x = self.pool1(x)

        # Layers 2 - 4
        x = F.relu(self.conv2(x))
        x = F.relu(self.conv3(x))
        x = F.relu(self.conv4(x))
        x = self.pool2(x)

        # Dropout & Flatten
        x = self.dropout(x)
        x = torch.flatten(x, start_dim=1)

        # Fully Connected Layers
        x = F.relu(self.fcn1(x))
        x = F.relu(self.fcn2(x))
        x = F.softmax(self.fcn3(x), dim=1)

        return x

# Quick verification
if __name__ == "__main__":
    model = CNNArchitecture(num_classes=10)
    dummy_input = torch.randn(32, 1, 28, 28)  # Batch size = 32
    output = model(dummy_input)
    print("Output shape:", output.shape)  # Expected: torch.Size([32, 10])
````
