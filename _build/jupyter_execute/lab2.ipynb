# Laboratory Task 2: Forward Pass & Error Computation

## Task Overview
Perform a single forward pass through a multi-layer neural network with ReLU activations and compute the prediction error.

---

## 1. Given Parameters

* **Input vector ($x$)**: $[1, 0, 1]^T$
* **Target output ($y$)**: $[1]$
* **Activation function**: $f(Z) = \max(0, Z)$ (ReLU)

### Weights and Biases
* **Hidden Unit Weights ($W_h$)**:
  $$\begin{bmatrix} 0.2 & -0.3 \\ 0.4 & 0.1 \\ -0.5 & 0.2 \end{bmatrix}$$
* **Output Unit Weights ($W_o$)**:
  $$\begin{bmatrix} -0.3 \\ -0.2 \end{bmatrix}$$
* **Biases ($\theta$)**: $\theta_1 = -0.4$, $\theta_2 = 0.2$, $\theta_3 = 0.1$

---

## 2. Calculation Steps

### Step 1: Hidden Layer Activations
$$Z_{h1} = (1 \cdot 0.2) + (0 \cdot 0.4) + (1 \cdot -0.5) + (-0.4) = -0.7$$
$$Z_{h2} = (1 \cdot -0.3) + (0 \cdot 0.1) + (1 \cdot 0.2) + 0.2 = 0.1$$

Applying ReLU:
$$A_{h1} = \max(0, -0.7) = 0$$
$$A_{h2} = \max(0, 0.1) = 0.1$$

### Step 2: Output Layer Activation
$$Z_o = (0 \cdot -0.3) + (0.1 \cdot -0.2) + 0.1 = 0.08$$
$$\hat{y} = \max(0, 0.08) = 0.08$$

### Step 3: Error Calculation
$$\text{Linear Error } (\delta) = \hat{y} - y = 0.08 - 1 = -0.92$$
$$\text{Squared Error } (E) = \frac{1}{2}(\hat{y} - y)^2 = 0.4232$$

---

## 3. Python Implementation

```python


````python
import numpy as np

# Inputs and Target
x = np.array([1, 0, 1])
y = 1.0

# Weights and Biases
W_h = np.array([[0.2, -0.3],
                [0.4, 0.1],
                [-0.5, 0.2]])
W_o = np.array([-0.3, -0.2])
theta_h = np.array([-0.4, 0.2])
theta_o = 0.1

# ReLU function
relu = lambda z: np.maximum(0, z)

# Forward pass - Hidden Layer
Z_h = np.dot(x, W_h) + theta_h
A_h = relu(Z_h)

# Forward pass - Output Layer
Z_o = np.dot(A_h, W_o) + theta_o
y_hat = relu(Z_o)

# Error
error = y_hat - y
squared_error = 0.5 * (error ** 2)

print(f"Predicted Output (y_hat): {y_hat}")
print(f"Linear Error: {error:.4f}")
print(f"Squared Error: {squared_error:.4f}")
````
